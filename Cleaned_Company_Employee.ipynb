# Cleaned Company Employee Dataset

**Task11** — Load, inspect, clean, verify, and export a messy employee dataset.

```python
import pandas as pd

df = pd.read_csv("Day11_Messy_Company_Employee_Dataset.csv")
print("Shape:", df.shape)
display(df.head())
print("\nData types:")
display(df.dtypes)
print("\nMissing values:")
display(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

## Cleaning approach

- Strip whitespace from text fields.
- Standardize known categorical variants.
- Use median imputation for missing numeric values.
- Use mode imputation for missing categorical values.
- Convert the joining date to datetime.
- Remove exact duplicate records with `drop_duplicates()`.

`dropna()` and forward filling are not used because dropping valid employee rows would lose information, while the dataset is not a time-series where forward filling is appropriate.

```python
# Clean whitespace and categorical inconsistencies
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)

category_maps = {
    "Department": {"Human Resources": "HR", "Information Technology": "IT", "Fin": "Finance"},
    "Gender": {"M": "Male", "m": "Male", "F": "Female", "f": "Female"},
    "Work_Mode": {"WFO": "Office", "On-site": "Office", "On Site": "Office", "WFH": "Remote"}
}
for col, mapping in category_maps.items():
    df[col] = df[col].replace(mapping)

# Numeric conversion + median imputation
numeric_cols = ["Age", "Annual_Salary", "Experience_Years", "Performance_Score"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

# Categorical mode imputation
for col in ["Department", "Gender", "City", "Work_Mode"]:
    df[col] = df[col].fillna(df[col].mode()[0])

# Date conversion
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], errors="coerce")

# Remove duplicates
df = df.drop_duplicates().reset_index(drop=True)

display(df.head())
```

```python
print("Final shape:", df.shape)
print("\nFinal missing values:")
display(df.isnull().sum())
print("\nFinal duplicate rows:", df.duplicated().sum())
print("\nFinal data types:")
display(df.dtypes)

df.to_csv("Cleaned_Company_Employee.csv", index=False)
print("\nSaved: Cleaned_Company_Employee.csv")
```
